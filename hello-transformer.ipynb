```python
from transformers import pipeline
import pandas as pd
```

```python
classifier = pipeline("text-classification", model = "michellejieli/emotion_text_classifier")
text = "Wow! What a lovely day."
outputs = classifier(text)
pd.DataFrame(outputs)
```
